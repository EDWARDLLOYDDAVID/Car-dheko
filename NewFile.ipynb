```python
import pandas as pd

bangalore_df = pd.read_excel('Dataset/bangalore_cars.xlsx')
jaipur_df =  pd.read_excel('Dataset/jaipur_cars.xlsx')
chennai_df =  pd.read_excel('Dataset/chennai_cars.xlsx')
hyderabad_df =  pd.read_excel('Dataset/hyderabad_cars.xlsx')
kolkata_df =  pd.read_excel('Dataset/kolkata_cars.xlsx')
delhi_df =  pd.read_excel('Dataset/delhi_cars.xlsx')
```

```python
# Adding the Column for the DataSet to mention the City

chennai_df['City'] = 'Chennai'
delhi_df['City'] = 'Delhi'
hyderabad_df['City'] = 'Hyderabad'
jaipur_df['City'] = 'Jaipur'
kolkata_df['City'] = 'Kolkata'
bangalore_df['City'] = 'Bangalore'
```

```python
jaipur_df
```

```python
df1 = pd.concat([bangalore_df,chennai_df],ignore_index = True)
df2 = pd.concat([delhi_df,hyderabad_df],ignore_index = True)
df3 = pd.concat([jaipur_df,kolkata_df], ignore_index = True)
df4 = pd.concat([df1,df2],ignore_index = True)
final_df = pd.concat([df4,df3],ignore_index = True)
final_df.shape
```
